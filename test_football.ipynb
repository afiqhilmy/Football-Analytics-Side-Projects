# Iniesta

```python
import matplotlib.pyplot as plt
from mplsoccer import Pitch
import seaborn as sns
from statsbombpy import sb
from matplotlib.patches import Rectangle

# --- 1. SET FUTURISTIC DARK THEME AND COLOR PALETTE ---
bg_color = '#0D0E15'
grid_color = '#1C1F2B'
text_main = '#FFFFFF'
text_muted = '#78909C'

# --- 2. FETCH EVENTS DATA ---
# 2008/09 Champions League Final (Barcelona 2-0 Man Utd)
events = sb.events(match_id=3750201)

# 2. Filter safely using case-insensitive partial string matching
iniesta_events = events[
    events['player'].str.contains('Iniesta', case=False, na=False) &
    events['location'].notna()
]

print(f"Total actions logged for Iniesta: {len(iniesta_events)}")

# 3. Extract x, y coordinates
x = [loc[0] for loc in iniesta_events['location']]
y = [loc[1] for loc in iniesta_events['location']]

# --- 5. DRAW PITCH WITH FUTURISTIC THEME ---
pitch = Pitch(
    pitch_type='statsbomb',
    pitch_color=bg_color,
    line_color=grid_color
)
fig, ax = pitch.draw(figsize=(10, 7))

# Set surrounding figure canvas color to match background
fig.patch.set_facecolor(bg_color)

# --- 6. PLOT HEATMAP ---
# Using Seaborn for smooth continuous blending with the 'rocket' palette
sns.kdeplot(
    x=x,
    y=y,
    ax=ax,
    fill=True,
    thresh=0.07,
    levels=100,
    cmap="rocket",
    bw_adjust=1.2,
    alpha=0.75,
    clip=((0, 120), (0, 80)),
    linewidths=0,
)

# --- 5. HARD RECTANGLE CROP AT PITCH LINES ---
pitch_boundary = Rectangle((0, 0), 120, 80, transform=ax.transData)
for collection in ax.collections:
    collection.set_clip_path(pitch_boundary)

# --- MAIN TITLE ---
fig.suptitle(
    "Andres Iniesta - Action Map",
    color=text_main,
    fontfamily='Open Sans',
    fontsize=20,
    fontweight='bold',
    y=0.98  # Controls vertical height of main title
)

# --- SUBTITLE (MUTED) ---
plt.title(
    "2008/09 UEFA Champions League Final vs Manchester United",
    color=text_muted,
    fontfamily='Open Sans',
    fontweight='bold',
    fontsize=15,
    pad=10  # Space between subtitle and pitch edge
)

# Subtitle / Attribution text using muted color
fig.text(
    0.0, 0.01,
    "Data Source: StatsBomb Open Data",
    ha='left',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

# Subtitle / Attribution text using muted color
fig.text(
    0.40, 0.01,
    "▶ ▶ ▶ ▶\nAttacking Direction",
    ha='center',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

fig.text(
    0.98, 0.01,
    "Visual Analysis by Afiq Hilmy | linkedin.com/in/afiqhilmy",
    ha='right',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

plt.show()
```

# Combine

```python
import matplotlib.pyplot as plt
from mplsoccer import Pitch
from statsbombpy import sb

# --- 1. SET FUTURISTIC DARK THEME AND COLOR PALETTE ---
bg_color = '#0D0E15'
grid_color = '#1C1F2B'
text_main = '#FFFFFF'
text_muted = '#78909C'

# --- 2. FETCH EVENTS DATA ---
# 2008/09 Champions League Final (Barcelona 2-0 Man Utd)
events = sb.events(match_id=3750201)

# Target players in order
players_info = [
    {"name": "Andrés Iniesta", "query": "Iniesta"},
    {"name": "Xavi", "query": "Xavier Hernández Creus"},
    {"name": "Sergio Busquets", "query": "Busquets"},
    {"name": "Lionel Messi", "query": "Messi"}
]

# --- 3. INITIALIZE VERTICAL PITCH ---
pitch = Pitch(
    pitch_type='statsbomb',
    pitch_color=bg_color,
    line_color=grid_color
)

# --- 4. CREATE SQUARE CANVAS WITH 1x4 SUBPLOTS ---
fig, axs = plt.subplots(1, 4, figsize=(14, 14), facecolor=bg_color)

# --- 5. PLOT EACH PLAYER ON VERTICAL PITCH ---
for ax, p in zip(axs, players_info):
    # Draw vertical pitch on current subplot
    pitch.draw(ax=ax)

    # Filter player events safely
    p_events = events[
        events['player'].str.contains(p['query'], case=False, na=False) &
        events['location'].notna()
    ]

    # Extract x and y coordinates
    x = [loc[0] for loc in p_events['location']]
    y = [loc[1] for loc in p_events['location']]

    # Plot KDE heatmap (mplsoccer handles coordinate conversion automatically)
    if len(x) > 0:
        pitch.kdeplot(
            x, y, ax=ax,
            fill=True,
            thresh=0.08,
            levels=100,
            cmap="rocket",
            bw_adjust=1.2,
            alpha=0.8,
            cut=0
        )

    # Set individual player title above pitch
    ax.set_title(p['name'], color=text_main, fontfamily='Open Sans', fontsize=15, fontweight='bold', pad=12)

# --- 6. ADD COMBINED HEADERS & FOOTERS ---
# Main Title
fig.suptitle(
    "FC Barcelona Key Players - Action Map Analysis",
    color=text_main,
    fontfamily='Open Sans',
    fontsize=20,
    fontweight='bold',
    y=0.76
)

# Subtitle
fig.text(
    0.50, 0.73,
    "2008/09 UEFA Champions League Final vs Manchester United",
    ha='center',
    color=text_muted,
    fontfamily='Open Sans',
    fontweight='bold',
    fontsize=13
)

# Footers
fig.text(
    0.05, 0.24,
    "Data Source: StatsBomb Open Data",
    ha='left',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

fig.text(
    0.50, 0.24,
    "▲ ▲ ▲ ▲\nAttacking Direction",
    ha='center',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

fig.text(
    0.95, 0.24,
    "Visual Analysis by Afiq Hilmy | linkedin.com/in/afiqhilmy",
    ha='right',
    fontfamily='Open Sans',
    fontweight='bold',
    color=text_muted,
    fontsize=9
)

# Adjust margins to center the vertical pitches cleanly inside the square layout
plt.subplots_adjust(top=0.70, bottom=0.28, left=0.05, right=0.95, wspace=0.15)

# Save or Display Image
plt.savefig("barca_action_maps_1x4.png", dpi=300, bbox_inches='tight', facecolor=bg_color)
plt.show()
```
