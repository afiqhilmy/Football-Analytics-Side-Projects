# 4. Per MP stats

```python
columns_to_normalize = [
    'accuratePasses',
    'assists',
    'goals',
    'ballRecovery',
    'bigChancesCreated',
    'expectedGoals',
    'expectedAssists',
    'groundDuelsWon',
    'interceptions',
    'keyPasses',
    'successfulDribbles',
    'tacklesWon',
    'accurateLongBalls'
]

# Ensure minutesPlayed is not zero to avoid division by zero errors.
# The previous filtering step (players_over_1000_minutes) already ensures this for a subset of data,
# but for filtered_data which is derived from all data, we should be cautious.
# For any rows where 'minutesPlayed' might be 0, we'll replace it with 1 before division
# to avoid NaNs or inf, as requested to replace values.
# However, given the context, all minutesPlayed values seem to be positive.

# Perform the division for each specified column
for col in columns_to_normalize:
    # Use .loc to avoid SettingWithCopyWarning and ensure changes are made to the original filtered_data
    # We'll handle potential division by zero by replacing 0 with NaN before division, then filling NaN with 0
    # This handles cases where 'minutesPlayed' might be 0 for some players, although not expected here.
    # More robust would be to filter out players with 0 minutesPlayed before this operation.
    # Given 'minutesPlayed' in filtered_data comes from original 'data' and likely contains non-zero values
    # for the relevant rows, direct division should be fine.
    filtered_data.loc[filtered_data['minutesPlayed'] != 0, col] = filtered_data.loc[filtered_data['minutesPlayed'] != 0, col] / filtered_data.loc[filtered_data['minutesPlayed'] != 0, 'minutesPlayed']
    # For players with 0 minutesPlayed, their stats per minute would also be 0
    filtered_data.loc[filtered_data['minutesPlayed'] == 0, col] = 0


# Display the first few rows of the modified DataFrame
filtered_data.head()
```

# 5. Per 90 Stats

```python
columns_to_multiply_by_90 = [
    'accuratePasses',
    'assists',
    'goals',
    'ballRecovery',
    'bigChancesCreated',
    'expectedGoals',
    'expectedAssists',
    'groundDuelsWon',
    'interceptions',
    'keyPasses',
    'successfulDribbles',
    'tacklesWon',
    'accurateLongBalls'
]

# Multiply each specified column by 90
for col in columns_to_multiply_by_90:
    # Using .loc to modify the DataFrame in place and avoid SettingWithCopyWarning
    filtered_data.loc[:, col] = filtered_data.loc[:, col] * 90

# Display the first few rows of the modified DataFrame
filtered_data.head()
```
